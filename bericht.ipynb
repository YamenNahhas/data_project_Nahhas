## Live-Abfrage aus dem Notebook

Der folgende Code ruft öffentliche Repository-Metadaten und Dateinamen erneut über die GitHub REST API ab. Für öffentliche Repositories ist kein Token erforderlich. Für private Repositories kann `GITHUB_TOKEN` als Umgebungsvariable gesetzt werden – niemals einen Token direkt in ein Notebook schreiben.

```python
import os
import requests

owner, repo = 'YamenNahhas', 'data_project_Nahhas'
headers = {'Accept': 'application/vnd.github+json'}
if os.getenv('GITHUB_TOKEN'):
    headers['Authorization'] = f"Bearer {os.environ['GITHUB_TOKEN']}"

base = f'https://api.github.com/repos/{owner}/{repo}'
meta_response = requests.get(base, headers=headers, timeout=20)
meta_response.raise_for_status()
meta = meta_response.json()
files_response = requests.get(f'{base}/contents', headers=headers, timeout=20)
files_response.raise_for_status()

print('Repository:', meta['full_name'])
print('Standard-Branch:', meta['default_branch'])
print('Root-Dateien:', [item['name'] for item in files_response.json()])
```

## Weitere praktische Abfragen

- Commits anzeigen: `GET /repos/{owner}/{repo}/commits`
- Issues auflisten: `GET /repos/{owner}/{repo}/issues`
- Branches abrufen: `GET /repos/{owner}/{repo}/branches`
- Dateiinhalt lesen: `GET /repos/{owner}/{repo}/contents/{path}`

Bei schreibenden API-Aufrufen (z. B. Datei erstellen) muss ein Token über eine sichere Umgebungsvariable mit passenden Rechten verwendet werden.
